#### 16. 如何给已有数组添加一圈值为 0 的边框？ （★☆☆）


````python
Z = np.ones((5,5))
print(Z)
 
Z[:, [0, -1]] = 0
Z[[0, -1], :] = 0
print(Z)
````

<details><summary>第 16 题参考答案（先独立完成再查看）</summary>

```python
Z = np.ones((5,5))
Z = np.pad(Z, pad_width=1, mode='constant', constant_values=0)
print(Z)

np.pad() 的作用是给数组的边缘“加一圈”
Z：要填充的数组。
pad_width=1：每个方向都加 1 层。二维数组就是上、下、左、右各加一层，所以 (5, 5) 变成 (7, 7)。
mode='constant'：新加的位置填固定值。
constant_values=0：这个固定值是 0。

# Not a solution to this problem but good to know: using fancy indexing for in-place edit
Z[:, [0, -1]] = 0
Z[[0, -1], :] = 0
print(Z)
```

#### 35. 如何原地计算 `((A+B)*(-A/2))`，尽量不创建副本？ （★★☆）


````python
A = np.ones(3)*1
B = np.ones(3)*2
np.add(A,B,out=B)#B=A+B返回B
np.divide(A,2,out=A)#A=A/2返回A
np.negative(A,out=A)#A=-A/2
np.multiply(A,B,out=A)#A=(-A/2*(A+B))
````

def rolling(a, window):

#### 78. 给定二维线段端点数组 `P0`、`P1` 和点 `p`，求 `p` 到每条直线的距离 （★★★）


````python
P0 = np.random.uniform(-10,10,(10,2))
P1 = np.random.uniform(-10,10,(10,2))
p  = np.random.uniform(-10,10,( 1,2))

def Distance(P0,P1,p):

    d = np.zeros(10)
    
    Pv = P1 - P0
    Pv_ = p - P0

    d_Pv = np.linalg.norm(Pv, axis=1)
    d_Pv_ = np.linalg.norm(Pv_, axis=1)

    

    for i in range(10):
        if d_Pv[i]==0:
            d[i] = d_Pv_[i]
        elif d_Pv_[i]==0:
            d[i] = 0
        else: 
            cos = np.dot(Pv[i],Pv_[i]) / ( d_Pv[i] * d_Pv_[i])
            sin = np.sqrt(1 - cos**2)
            d[i] = np.sqrt(Pv_[i,0]**2 + Pv_[i,1]**2) * sin

    return(d)
d = Distance(P0,P1,p)
print(d) 
    
````

def distance_points_to_lines(p: np.ndarray, p_1: np.ndarray, p_2: np.ndarray) -> np.ndarray:
    x_0, y_0 = p.T  # Shape -> (n points, )
    x_1, y_1 = p_1.T  # Shape -> (n lines, )
    x_2, y_2 = p_2.T  # Shape -> (n lines, )

    # Displacement vector coordinates from p_1 -> p_2
    dx = x_2 - x_1  # Shape -> (n lines, )
    dy = y_2 - y_1  # Shape -> (n lines, )

    # The 'cross product' term
    cross_term = x_2 * y_1 - y_2 * x_1  # Shape -> (n lines, )

    # Broadcast x_0, y_0 (n points, 1) and dx, dy, cross_term (1, n lines) -> (n points, n lines)
    numerator = np.abs(
        dy[np.newaxis, :] * x_0[:, np.newaxis]
        - dx[np.newaxis, :] * y_0[:, np.newaxis]
        + cross_term[np.newaxis, :]
    )
    denominator = np.sqrt(dx**2 + dy**2)  # Shape -> (n lines, )

    # Shape (n points, n lines) / (1, n_lines) -> (n points, n lines)
    return numerator / denominator[np.newaxis, :]
