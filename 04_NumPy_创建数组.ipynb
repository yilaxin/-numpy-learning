# 第四课：NumPy 创建数组

本课依据菜鸟教程的 [NumPy 创建数组](https://www.runoob.com/numpy/numpy-array-creation.html) 整理，重点学习 `empty`、`zeros`、`ones`、`full` 及其 `_like` 版本。

```python
import numpy as np
print("NumPy 版本：", np.__version__)
```

## 1. `zeros()` 与 `ones()`

它们接收形状 `shape`，创建全 0 或全 1 数组。`shape` 是整数时创建一维数组，是元组时创建多维数组。默认 `dtype` 通常是 `float64`。

```python
a = np.zeros(5)
b = np.zeros((2, 3), dtype=np.int32)
c = np.ones((2, 2), dtype=np.float32)
print(a, a.dtype)
print(b, b.dtype)
print(c, c.dtype)
```

## 2. `full()`：用指定值填充

`np.full(shape, fill_value)` 比先创建数组再逐项赋值更直接。

```python
sevens = np.full((2, 4), 7)
pi_grid = np.full((2, 3), 3.14, dtype=np.float32)
print(sevens)
print(pi_grid, pi_grid.dtype)
```

## 3. `empty()`：只分配内存

`np.empty()` 不会把数组初始化成 0，里面可能是旧内存残留的任意值。它适合马上会覆盖全部元素的场景；不要依赖它刚创建时的内容。

uninitialized[:] = 9 是 切片赋值。
这里的 : 表示“这一维中从头到尾的全部元素”：

a[0, :] = 1   # 第 0 行全部设为 1
a[:, 1] = 2   # 第 1 列全部设为 2
a[0:2, 1:3] = 8  # 指定区域全设为 8   0-1行的1-2列

和matlab一样，注意要-1



```python
uninitialized = np.empty((2, 3))
print("未初始化内容（每次可能不同）：\n", uninitialized)
uninitialized[:, :] = 9
print("全部覆盖后：\n", uninitialized)
```

## 4. `_like` 系列：沿用已有数组的属性

`zeros_like()`、`ones_like()`、`full_like()` 和 `empty_like()` 默认沿用参考数组的形状和数据类型。

```python
template = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int16)
print("模板：\n", template)
print("zeros_like：\n", np.zeros_like(template))
print("ones_like：\n", np.ones_like(template))
print("full_like：\n", np.full_like(template, 8))
```

注意：`full_like()` 会沿用模板的 `dtype`。如果模板是整数，填入小数可能被转换成整数。需要时显式指定 `dtype`。

```python
integer_template = np.array([1, 2, 3], dtype=np.int32)
print(np.full_like(integer_template, 3.14))
print(np.full_like(integer_template, 3.14, dtype=np.float64))
```

## 5. 补充：单位矩阵 `eye()` 与 `identity()`

单位矩阵的主对角线为 1，其余位置为 0。`eye()` 可以创建非方阵并移动对角线，功能更灵活。

```python
print("3×3 单位矩阵：\n", np.eye(3, dtype=np.int32))
print("3×4，向右偏移一条对角线：\n", np.eye(3, 4, k=1, dtype=np.int32))
print("identity：\n", np.identity(3))
 
```

## 6. 了解 `order`

许多创建函数支持 `order='C'`（按行优先）或 `order='F'`（按列优先）。数组显示结果相同，但内存布局不同。普通学习和大多数应用保持默认的 `'C'` 即可。

```python
c_order = np.ones((2, 3), order="C")
f_order = np.ones((2, 3), order="F")
print("C 数组：C连续=", c_order.flags.c_contiguous, "F连续=", c_order.flags.f_contiguous)
print("F 数组：C连续=", f_order.flags.c_contiguous, "F连续=", f_order.flags.f_contiguous)
```

## 如何选择？

| 需求 | 推荐函数 |
|---|---|
| 全部填 0 | `np.zeros()` |
| 全部填 1 | `np.ones()` |
| 全部填指定值 | `np.full()` |
| 马上覆盖全部元素，追求少一步初始化 | `np.empty()` |
| 沿用另一个数组的形状和类型 | `*_like()` |
| 创建单位矩阵 | `np.eye()` |
