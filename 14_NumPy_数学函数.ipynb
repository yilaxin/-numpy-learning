# 第十四课：NumPy 数学函数

本课依据菜鸟教程的 [NumPy 数学函数](https://www.runoob.com/numpy/numpy-mathematical-functions.html) 整理，重点学习三角函数、角度转换、舍入、绝对值、开方、指数和对数。

```python
import numpy as np
print("NumPy 版本：", np.__version__)
```

## 1. 三角函数使用弧度

`sin()`、`cos()`、`tan()` 的输入是弧度，不是角度。使用 `np.deg2rad()` 或 `np.radians()` 将角度转换为弧度。

```python
degrees = np.array([0, 30, 45, 60, 90])
radians = np.deg2rad(degrees)
print("弧度：", radians)
print("sin：", np.sin(radians))
print("cos：", np.cos(radians))
print("tan（不含 90°）：", np.tan(radians[:-1]))
```

浮点计算存在误差，例如 `cos(90°)` 可能得到接近 0 的小数而不是精确的 0。比较时使用 `np.isclose()`。

```python
cos_90 = np.cos(np.deg2rad(90))
print("cos(90°)：", cos_90)
print("是否接近 0：", np.isclose(cos_90, 0))
```

## 2. 反三角函数

`arcsin()`、`arccos()`、`arctan()` 返回弧度。使用 `rad2deg()` 或 `degrees()` 转回角度。

```python
sin_values = np.array([0, 0.5, np.sqrt(2) / 2, 1])
angles = np.rad2deg(np.arcsin(sin_values))
print("反正弦对应角度：", angles)
print("arctan2(1, 1) 的角度：", np.degrees(np.arctan2(1, 1)))#用于计算坐标轴上离x轴的角度，先y后x
```

## 3. 舍入函数

- `round()` / `around()`：舍入到指定小数位
- `floor()`：向负无穷方向取整
- `ceil()`：向正无穷方向取整
- `trunc()`：直接截去小数部分，朝 0 取整

负数最能体现三者区别。

```python
values = np.array([-1.7, -0.2, 0.6, 1.5, 5.55, 25.532])
print("round：", np.round(values, 1))#会四舍五入
print("floor：", np.floor(values))
print("ceil：", np.ceil(values))
print("trunc：", np.trunc(values))
```

NumPy 的舍入遵循“舍入到最近值，正好一半时趋向偶数”的规则，并且仍受浮点表示影响。不要把它简单理解成所有 `.5` 都远离 0。

-decimals=2   # 保留 2 位小数

-decimals=0   # 四舍五入到整数

-decimals=-1  # 四舍五入到十位

-decimals=-2  # 四舍五入到百位

```python
halves = np.array([0.5, 1.5, 2.5, 3.5, -0.5, -1.5])
print(np.round(halves))
print("负 decimals：", np.round([123, 156, 185], decimals=-1))
```

## 4. 绝对值和平方根

`np.abs()` 也支持复数，返回复数模长。实数负值直接开平方会得到 `nan`；需要复数结果时先转换为复数 dtype。

```python
numbers = np.array([-9, -4, 0, 4, 9])
print("绝对值：", np.abs(numbers))
print("非负数平方根：", np.sqrt(numbers[numbers >= 0]))
complex_numbers = np.array([-4, -1, 3 + 4j], dtype=np.complex128)
print("复数平方根：", np.sqrt(complex_numbers))
print("复数模长：", np.abs(complex_numbers))
```

## 5. 指数和对数

- `exp(x)`：$e^x$
- `exp2(x)`：$2^x$
- `log(x)`：自然对数
- `log2(x)`、`log10(x)`：以 2、10 为底
- `log1p(x)`：计算 $\log(1+x)$，对很小的 x 更精确
- `expm1(x)`：计算 $e^x-1$，对很小的 x 更精确

```python
x = np.array([0.0, 1.0, 2.0, 3.0])
print("exp：", np.exp(x))
print("exp2：", np.exp2(x))
positive = np.array([1.0, 2.0, 10.0, 100.0])
print("log：", np.log(positive))
print("log2：", np.log2(positive))
print("log10：", np.log10(positive))
```

实数对数要求输入大于 0。0 的对数是负无穷，负数的实数对数是 `nan`。需要复数结果时使用复数 dtype；需要忽略已知警告时可局部使用 `np.errstate()`。

```python
with np.errstate(divide="ignore", invalid="ignore"):#忽略警告
    real_result = np.log(np.array([-1.0, 0.0, 1.0]))
    complex_result = np.log(np.array([-1.0, 0.0 + 0j, 1.0], dtype=np.complex128))
print("实数结果：", real_result)
print("复数结果：", complex_result)
```

## 6. 限制范围：`clip()`

`np.clip(a, min, max)` 把小于下限的值设为下限，大于上限的值设为上限。常用于图像、传感器和模型输出。

```python
sensor = np.array([-5.0, 12.0, 55.0, 101.0])
print(np.clip(sensor, 0, 100))
```

## 7. 补充：距离与符号

`hypot(x, y)` 稳定计算 $\sqrt{x^2+y^2}$；`sign()` 返回负、零、正的符号；`copysign()` 把一个数的符号复制给另一个数。

```python
x = np.array([3.0, 5.0, 8.0])
y = np.array([4.0, 12.0, 15.0])
print("斜边：", np.hypot(x, y))
print("符号：", np.sign([-3, 0, 8]))
```

## 本课小结

- 三角函数输入弧度，使用 `deg2rad` 和 `rad2deg` 转换。
- 浮点结果比较使用 `isclose()`。
- `floor`、`ceil`、`trunc` 对负数的行为不同。
- `abs`、`sqrt`、`exp`、`log` 都逐元素运算。
- 注意平方根、对数等函数的定义域。
- `clip` 可将数值限制在给定区间。

## 本节练习

1. 生成 0° 到 360° 的 9 个等距角度，计算正弦和余弦。
2. 验证第一个与最后一个正弦值近似相等。
3. 对 `measurements` 分别做向下、向上和截断取整。
4. 把它限制在 `[0, 100]`。
5. 对正值计算自然对数。

```python
measurements = np.array([-3.7, 0.2, 18.9, 99.6, 105.4])
# 请在这里完成练习
deg_ =  np.linspace(0,360,9)
print(deg_)
deg_rad = np.deg2rad(deg_)
print(np.sin(deg_rad))
print(np.cos(deg_rad))
 
print(np.isclose(np.sin(deg_rad[0]),np.sin(deg_rad[-1])))
print(np.floor(measurements))
print(np.ceil(measurements))
print(np.trunc(measurements))
print(np.clip(measurements,0,100))
print(np.log(measurements[measurements>0]))
```

## 参考答案

```python
angles = np.linspace(0, 360, 9)
radians = np.deg2rad(angles)
sines = np.sin(radians)
cosines = np.cos(radians)
print("角度：", angles)
print("sin：", sines)
print("cos：", cosines)
print("首尾 sin 接近：", np.isclose(sines[0], sines[-1]))
print("floor：", np.floor(measurements))
print("ceil：", np.ceil(measurements))
print("trunc：", np.trunc(measurements))
clipped = np.clip(measurements, 0, 100)
print("限制范围：", clipped)
positive = measurements[measurements > 0]
print("正值的自然对数：", np.log(positive))
```
