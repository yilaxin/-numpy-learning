# 第八课：NumPy 高级索引

本课依据菜鸟教程的 [NumPy 高级索引](https://www.runoob.com/numpy/numpy-advanced-indexing.html) 整理，重点学习整数数组索引、花式索引、布尔索引与 `np.ix_()`。

高级索引能够选择不连续位置或按条件筛选数据。与基本切片不同，高级索引的结果通常是副本。

```python
import numpy as np
print("NumPy 版本：", np.__version__)
```

## 1. 整数数组索引

把多个索引放在列表或数组中，可以一次取得多个不连续位置，也可以改变顺序或重复选择。

```python
a = np.arange(10, 20)
print("原数组：", a)
print("位置 0、3、7：", a[[0, 3, 7]])
print("倒序挑选：", a[[4, 2, 0]])
print("重复选择：", a[[1, 1, 5]])
print("负索引：", a[[-1, -3]])
```

二维数组用一个整数数组索引时，默认选择第 0 轴上的行。

```python
m = np.arange(32).reshape(8, 4)
print("原数组：\n", m)
print("选择第 5、3、2、8 行：\n", m[[4, 2, 1, 7]])
```

## 2. 用行、列索引成对取元素

`x[rows, cols]` 会把 `rows` 与 `cols` 对应位置配对，而不是取得所有行列组合。例如行 `[0,1,2]`、列 `[0,1,0]` 表示坐标 `(0,0)`、`(1,1)`、`(2,0)`。

```python
x = np.array([[1, 2], [3, 4], [5, 6]])
rows = np.array([0, 1, 2])
cols = np.array([0, 1, 0])
print(x)
print("坐标对应元素：", x[rows, cols])
```

行列索引本身也可以是二维，从而让结果保留指定形状。

```python
grid = np.arange(12).reshape(4, 3)
row_indices = np.array([[0, 0], [3, 3],[0,0]])
col_indices = np.array([[0, 2], [0, 2],[0,1]])#合并之后和自己的形状相同，比如这里就是3，2
print("网格：\n", grid)
print("四个角：\n", grid[row_indices, col_indices])
```

## 3. `np.ix_()`：取得行列的笛卡尔积

如果想选择“这些行与这些列组成的整个子矩阵”，不能直接写 `m[rows, cols]`，因为那会成对取值。可以使用 `np.ix_(rows, cols)`。

```python
m = np.arange(20).reshape(4, 5)
rows = [0, 2, 3]
cols = [1, 4]
print("原数组：\n", m)
 
print("指定行列的全部组合：\n", m[np.ix_(rows, cols)])#本质上就是指定行列取子矩阵
 
 
```

## 4. 布尔索引

比较运算会产生布尔数组。把它用作索引，即可取得所有 `True` 位置的元素。筛选结果通常会变成一维数组。

```python
values = np.array([[2, 7, 4], [9, 1, 8], [3, 6, 5]])
mask = values > 5
print("数组：\n", values)
print("条件掩码：\n", mask)
print("大于 5 的元素：", values[mask])
print("等价写法：", values[values > 5])
```

组合多个条件时，每个条件要加括号，并使用逐元素运算符 `&`（且）、`|`（或）、`~`（非），不能使用 Python 的 `and`、`or`、`not`。

```python
numbers = np.arange(1, 21)
selected = numbers[(numbers >= 5) & (numbers <= 15) & (numbers % 2 == 0)]
outside = numbers[(numbers < 5) | (numbers > 15)]
print("5 到 15 之间的偶数：", selected)
print("区间之外：", outside)
```

## 5. 过滤缺失值 `NaN`

`NaN` 不能用 `x == np.nan` 正确筛选。应使用 `np.isnan()`；前面加 `~` 表示取反，即保留不是 `NaN` 的元素。

```python
measurements = np.array([1.5, np.nan, 2.7, np.nan, 4.1])
print("NaN 位置：", np.isnan(measurements))
print("有效数据：", measurements[~np.isnan(measurements)])
```

布尔条件也可以选择整行。例如，根据每行总分选择学生。

```python
scores = np.array([[80, 90, 70], [95, 92, 98], [60, 75, 68], [88, 85, 90]])
print(scores)
totals = scores.sum(axis=1)
high_score_rows = scores[totals >= 260]
print("总分：", totals)
print("总分至少 260 的行：\n", high_score_rows)
```

## 6. 用高级索引修改数组

高级索引可以出现在赋值左侧，对指定位置批量更新。下面把负数替换为 0。

```python
data = np.array([3, -2, 7, -5, 1])
data[data < 0] = 0
print(data)
```

## 7. 高级索引通常返回副本

与基本切片返回视图不同，花式索引和布尔索引取得的结果通常不与原数组共享内存。修改结果不会影响原数组。

```python
original = np.arange(8)
basic_slice = original[1:4]
fancy_result = original[[1, 2, 3]]
print("切片共享内存：", np.shares_memory(original, basic_slice))
print("高级索引共享内存：", np.shares_memory(original, fancy_result))
fancy_result[0] = 999
print("修改高级索引结果后的原数组：", original)
```

## 8. 补充：`np.where()`

`np.where(condition, x, y)` 可以按条件在两个值或数组之间选择，并保持原数组形状。只有一个参数时，`np.where(condition)` 返回满足条件的坐标。

```python
values = np.array([-3, 2, -1, 5, 0])
print("负数替换为 0：", np.where(values < 0, 0, values))#True选前面的，false选后面的
print("正数的位置：", np.where(values > 0)[0])#因为返回的是只有一个元素的元组
```

## 本课小结

- 整数数组可以选择、重排或重复元素。
- `x[rows, cols]` 按坐标成对取值。
- `np.ix_()` 取得指定行列的全部组合。
- 布尔索引按条件筛选，组合条件使用 `&`、`|`、`~`。
- 高级索引结果通常是副本，基本切片通常是视图。
- `np.where()` 可以保持形状地按条件选值。

## 本节练习

对下面的成绩表完成：

1. 选择第 1、3、5 名学生的全部成绩。
2. 取得坐标 `(0,0)`、`(2,1)`、`(4,2)` 的三个成绩。
3. 选择第 2、4 名学生的第 1、3 门成绩组成 `2×2` 子矩阵。
4. 筛选所有大于等于 90 的成绩。
5. 将所有低于 60 的成绩改为 60。

```python
exam_scores = np.array([
    [88, 76, 95],
    [59, 82, 73],
    [91, 94, 89],
    [67, 58, 72],
    [96, 87, 92],
])
# 请在这里完成练习
print(exam_scores[[0,2,4]])
print(exam_scores[[0,2,4],[0,1,2]])
print(exam_scores[np.ix_([1,3],[0,2])])
print(exam_scores[exam_scores>90])
exam_scores[exam_scores<60] = 60
print(exam_scores)
```

## 参考答案

```python
print("第 1、3、5 名：\n", exam_scores[[0, 2, 4]])
print("三个坐标：", exam_scores[[0, 2, 4], [0, 1, 2]])
print("2×2 子矩阵：\n", exam_scores[np.ix_([1, 3], [0, 2])])
print("至少 90 分：", exam_scores[exam_scores >= 90])
adjusted = exam_scores.copy()
adjusted[adjusted < 60] = 60
print("调整后：\n", adjusted)
```
