# 第六课：NumPy 从数值范围创建数组

本课依据菜鸟教程的 [NumPy 从数值范围创建数组](https://www.runoob.com/numpy/numpy-array-from-numerical-ranges.html) 整理，学习 `arange()`、`linspace()` 和 `logspace()`。

```python
import numpy as np
print("NumPy 版本：", np.__version__)
```

## 1. `arange()`：按步长生成

常用形式：`np.arange(start, stop, step, dtype=None)`。规则与 Python 的 `range()` 相似：包含 `start`，不包含 `stop`。区别是 `arange()` 返回数组，并且可以使用浮点数。

```python
print(np.arange(5))
print(np.arange(10, 20, 2))
print(np.arange(5, 0, -1))#倒着生成也不包含stop
print(np.arange(5, dtype=np.float64))
```

### 浮点步长的注意事项

浮点数存在表示误差，使用浮点步长时元素数量和末端值有时不直观。需要明确生成固定数量的浮点样本时，优先使用 `linspace()`。

```python
x = np.arange(0, 1, 0.1)
print(x)
print("元素数量：", x.size)
print("最后一个元素：", repr(x[-1]))
```

## 2. `linspace()`：按数量等分

`np.linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None)` 在起点和终点之间生成指定数量的等距样本。默认包含终点。

```python
print(np.linspace(0, 1, 5))#默认包含结束点
print(np.linspace(0, 1, 5, endpoint=False))#不包含结束点
values, step = np.linspace(10, 20, 6, retstep=True)#一起返回步长
print("数组：", values)
print("间距：", step)
```

## 3. `arange()` 与 `linspace()` 的区别

| 函数 | 你指定什么 | `stop` 默认是否包含 | 典型用途 |
|---|---|---|---|
| `arange()` | 步长 `step` | 不包含 | 整数序列、已知步长 |
| `linspace()` | 元素数量 `num` | 包含 | 绘图采样、固定数量等分点 |

```python
by_step = np.arange(0, 11, 2)
by_count = np.linspace(0, 10, 6 )
print("arange：", by_step)
print("linspace：", by_count)
print("数值近似相同：", np.allclose(by_step, by_count))
#allclose直接判断所有元素，返回一个bool值
#isclose逐个判断，返回一个bool数组
```

## 4. `logspace()`：按指数等分

`np.logspace(start, stop, num=50, base=10.0)` 先在指数上等距取点，再计算 `base ** 指数`。例如 `start=0, stop=3, base=10` 对应从 $10^0$ 到 $10^3$。
是有默认值的

```python
powers_of_ten = np.logspace(0, 3, 4)
powers_of_two = np.logspace(0, 9, 10, base=2)
print("10 的幂：", powers_of_ten)
print("2 的幂：", powers_of_two)
```

## 5. 补充：`geomspace()`

`geomspace(start, stop, num)` 直接指定首尾数值，生成等比数列；`logspace()` 指定的是指数范围。两者都适合跨越多个数量级的数据。

```python
print(np.geomspace(1, 1000, 4))
print(np.geomspace(1, 256, 9))
```

## 6. 与 `reshape()` 组合

范围函数创建的都是一维数组，可以再用 `reshape()` 整理成多维结构。

```python
grid = np.arange(1, 13).reshape(3, 4)
column = np.linspace(0, 1, 5).reshape(-1, 1)
print("网格：\n", grid)
print("列向量：\n", column)
```

## 7. 实际应用：生成采样点

绘制函数时经常用 `linspace()` 生成固定数量的采样点。下面在 $0$ 到 $2\pi$ 之间生成 9 个点并计算正弦值。

```python
angles = np.linspace(0, 2 * np.pi, 9)
sines = np.sin(angles)
for angle, sine in zip(angles, sines):
    print(f"角度={angle:6.3f}，sin={sine:6.3f}")#.3f表示保留三位小数，6表示总宽度为6，右对齐
```

## 本课小结

- 已知步长，用 `arange()`。
- 已知点数，用 `linspace()`。
- 按数量级生成，用 `logspace()` 或 `geomspace()`。
- 浮点范围通常优先使用 `linspace()`，避免步长累计误差。
- `endpoint=False` 可以排除终点。
