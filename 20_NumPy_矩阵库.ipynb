# 第 20 课：NumPy 矩阵库（Matrix）

`numpy.matlib` 提供传统矩阵对象。现在的 NumPy 编程更推荐使用二维 `ndarray` 和 `@` 运算符，但了解 `matlib` 有助于阅读旧代码。

```python
import numpy as np
import numpy.matlib as matlib
```

## 1. 创建矩阵和转置

`matlib.matrix` 或 `np.matrix` 创建矩阵对象，`T` 返回转置。`matlib.zeros`、`ones`、`eye` 可创建常用矩阵。

```python
m = matlib.matrix([[1, 2], [3, 4]])
print('矩阵：\n', m)
print('类型：', type(m))
print('转置：\n', m.T)
print('零矩阵：\n', matlib.zeros((2, 3)))
print('单位矩阵：\n', matlib.eye(3,k=1))
```

## 2. 矩阵乘法与 `ndarray` 对比

矩阵对象使用 `*` 表示矩阵乘法；二维 `ndarray` 中 `*` 是逐元素乘法，矩阵乘法应使用 `@` 或 `np.matmul`。为了避免歧义，推荐新代码统一使用 `ndarray` 和 `@`。

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print('ndarray 逐元素乘法：\n', a * b)
print('ndarray 矩阵乘法：\n', a @ b)
print('matrix 的 *：\n', matlib.matrix(a) * matlib.matrix(b))
```

## 3. 从字符串创建矩阵

`mat` 可以把字符串解析成矩阵，行用分号分隔，列用空格分隔。它适合快速输入小型示例，但生产代码更建议使用 `np.array`。

```python
text_matrix = np.asmatrix('1 2; 3 4')
print(text_matrix)
print('行列式：', np.linalg.det(np.asarray(text_matrix)))
print('转为 ndarray：', np.asarray(text_matrix))
```

## 本节练习

创建两个 2×2 的 `ndarray`，分别计算逐元素乘法、矩阵乘法和第一个矩阵的转置；再用 `matlib.eye(3)` 创建 3×3 单位矩阵。

```python
x = np.arange(4).reshape(2,2)
y = np.arange(4,8).reshape(2,2)
print(x)
print(y)
print(x * y)
print(matlib.matrix(x)*matlib.matrix(y))
print(x.T)
print(matlib.eye(3))
```

```python
x = np.array([[0, 1], [2, 3]])
y = np.array([[4, 5], [6, 7]])
print('逐元素乘法：\n', x * y)
print('矩阵乘法：\n', x @ y)
print('x 的转置：\n', x.T)
print('单位矩阵：\n', matlib.eye(3))
```
