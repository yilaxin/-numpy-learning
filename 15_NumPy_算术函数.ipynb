本课依据菜鸟教程的 [NumPy 算术函数](https://www.runoob.com/numpy/numpy-arithmetic-operations.html) 整理，学习逐元素加减乘除、倒数、乘方、整除、余数以及通用函数的实用参数。

```python
import numpy as np
print("NumPy 版本：", np.__version__)
```

## 1. 基本逐元素运算

运算符与函数一一对应：`+` / `add()`、`-` / `subtract()`、`*` / `multiply()`、`/` / `divide()`。数组形状必须相同或满足广播规则。

```python
a = np.arange(1, 10, dtype=np.float64).reshape(3, 3)
print(a)
b = np.array([10.0, 20.0, 30.0])
print("a + b：\n", np.add(a, b))
print("a - b：\n", np.subtract(a, b))
print("a * b：\n", np.multiply(a, b))
print("a / b：\n", np.divide(a, b))
```

## 2. `*` 不是矩阵乘法

`a * b` 是对应元素相乘。矩阵乘法使用 `a @ b` 或 `np.matmul(a, b)`。一维向量点积也可用 `np.dot()`。

```python
left = np.array([[1, 2], [3, 4]])
right = np.array([[5, 6], [7, 8]])
print("逐元素乘法：\n", left * right)
print("矩阵乘法：\n", left @ right)
print("向量点积：", np.dot([1, 2, 3], [4, 5, 6]))
```

## 3. 三种除法

- `/` 或 `true_divide()`：真除法，通常得到浮点数
- `//` 或 `floor_divide()`：向下整除
- `%`、`mod()` 或 `remainder()`：余数

对于负数，整除朝负无穷方向取整，所以结果可能与“截去小数”不同。

```python
numerators = np.array([10, 20, -10, -20])
denominators = np.array([3, 6, 3, 6])
print("真除法：", np.true_divide(numerators, denominators))
print("向下整除：", np.floor_divide(numerators, denominators))
print("余数：", np.remainder(numerators, denominators))
quotient, remainder = np.divmod(numerators, denominators)#同时得到向下整除加求余数
print("divmod：", quotient, remainder)
print("验证 a = q*b+r：", np.array_equal(numerators, quotient * denominators + remainder))
```

## 4. 倒数 `reciprocal()`

整数数组的倒数仍按整数 dtype 计算，例如 `1/2` 会得到 0。要获得小数倒数，应先使用浮点类型。0 的倒数会产生无穷或警告。

```python
integers = np.array([1, 2, 4])
floats = integers.astype(np.float64)
print("整数倒数：", np.reciprocal(integers))#由于为int型，所以会截掉小数
print("浮点倒数：", np.reciprocal(floats))
```

## 5. 幂运算

`np.power(base, exponent)` 或 `base ** exponent` 逐元素计算，并支持广播。整数底数的负整数指数无法用整数 dtype 表示，应转为浮点数。

```python
bases = np.array([2.0, 3.0, 10.0])
exponents = np.array([3, 2, -1])
print(np.power(bases, exponents))
print(bases ** exponents)
matrix = np.arange(1, 7).reshape(2, 3)
print("每列使用不同指数：\n", matrix ** np.array([1, 2, 3]))
```

## 6. 逐元素比较函数

`maximum()` 与 `minimum()` 在两个数组间逐元素取较大/较小值。`fmax()`、`fmin()` 在遇到 `NaN` 时倾向于选择非 `NaN` 值。它们不同于对单个数组做聚合的 `max()`、`min()`。

```python
x = np.array([1.0, 8.0, np.nan, 4.0])
y = np.array([2.0, 3.0, 5.0, np.nan])
print("maximum：", np.maximum(x, y))
print("fmax：", np.fmax(x, y))
print("minimum：", np.minimum(x, y))
print("fmin：", np.fmin(x, y))
```

## 7. 除以零与 `errstate()`

浮点数除以零通常产生 `inf` 或 `nan` 并给出警告。可以用 `np.errstate()` 在局部明确处理，再用 `isfinite()` 筛选有效结果。不要在全局静默所有数值错误。

```python
values = np.array([10.0, 0.0, -5.0])
divisors = np.array([2.0, 0.0, 0.0])
with np.errstate(divide="ignore", invalid="ignore"):
    result = values / divisors
print("结果：", result)
print("有限值：", result[np.isfinite(result)])
```

## 8. 通用函数的 `out` 与 `where`

NumPy 的许多逐元素函数是 ufunc。`out` 可指定输出数组，`where` 可控制哪些位置执行。使用 `where` 时最好提供已初始化的 `out`，避免未计算位置保留未初始化值。

```python
numerator = np.array([10.0, 20.0, 30.0])
denominator = np.array([2.0, 0.0, 5.0])
output = np.full(3, np.nan)
print(output)
np.divide(numerator, denominator, out=output, where=denominator != 0)
print(output)
```

## 9. 原地运算与类型限制

`a += b` 尝试写回原数组，节省一次分配，但结果必须能转换回原 dtype。整数数组不能安全地原地加浮点数。非原地 `a + b` 则可创建提升类型的新数组。

```python
integers = np.array([1, 2, 3], dtype=np.int32)
promoted = integers + 0.5
print("非原地结果：", promoted, promoted.dtype)
floats = integers.astype(np.float64)
floats += 0.5
print("浮点原地运算：", floats)
```

## 本课小结

- 基本算术函数逐元素运算并支持广播。
- `*` 是逐元素乘法，`@` 是矩阵乘法。
- `/`、`//`、`%` 分别是真除法、向下整除、余数。
- 倒数和负指数计算要注意整数 dtype。
- `maximum` 是逐元素比较，`max` 是聚合。
- `errstate` 可局部处理浮点异常。
- ufunc 的 `out` 和 `where` 可控制输出和计算位置。

## 本节练习

对 `sales` 和 `costs` 完成：

1. 计算利润 `sales - costs`。
2. 计算利润率 `profit / sales`，销售额为 0 的位置填 `nan`。
3. 把利润限制为最低 0（使用 `maximum`）。
4. 计算每个元素除以 7 的商和余数。
5. 验证 `商 × 7 + 余数` 可以还原整数销售额。

```python
sales = np.array([[120.0, 0.0, 350.0], [80.0, 210.0, 490.0]])
costs = np.array([[90.0, 15.0, 280.0], [95.0, 160.0, 420.0]])
# 请在这里完成练习
print('利润',sales - costs)
divide_ = np.full((2,3),np.nan)
np.divide(sales - costs,sales,out = divide_ ,where = sales!= 0)
print('利润率',divide_)
print(np.maximum(sales - costs,0))
sales_ = np.float64(sales)
print()
shang , mod_ = np.divmod(sales,7)
print(shang )
print(mod_)
print(np.array_equal(sales_,shang * 7 + mod_))
```

## 参考答案

```python
profit = np.subtract(sales, costs)
margin = np.full(sales.shape, np.nan)
np.divide(profit, sales, out=margin, where=sales != 0)
non_negative_profit = np.maximum(profit, 0)
integer_sales = sales.astype(np.int64)
quotient, remainder = np.divmod(integer_sales, 7)
print("利润：\n", profit)
print("利润率：\n", margin)
print("非负利润：\n", non_negative_profit)
print("商：\n", quotient)
print("余数：\n", remainder)
print("还原成功：", np.array_equal(quotient * 7 + remainder, integer_sales))
```
