NumPy 负责生成和计算数据，Matplotlib 负责把数据绘制成图形。本课使用折线图、子图和直方图完成一个小型可视化。

```python
import numpy as np
import matplotlib.pyplot as plt

print('Matplotlib 已导入')
```

## 1. 绘制正弦曲线

`plot(x, y)` 绘制折线，`title` 设置标题，`xlabel`/`ylabel` 设置坐标轴标签，`legend` 显示图例。Notebook 中使用 `plt.show()` 显示图形。

```python
x = np.arange(0, 2 * np.pi, 0.05)
y = np.sin(x)
plt.figure(figsize=(7, 4))
plt.plot(x, y, label='sin(x)')
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']  # 或 ['SimHei']
plt.rcParams['axes.unicode_minus'] = False             
plt.title('正弦曲线')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid(True)
plt.legend()
plt.show()
```

## 2. 子图比较正弦和余弦

`plt.subplot(行数, 列数, 编号)` 将画布分成多个区域。`tight_layout()` 可以减少标题和标签重叠。

`plt.tight_layout()` 会自动调整一张图里各个子图之间的间距，尽量避免标题、坐标轴标签和刻度文字互相挤压或被裁掉。

```python
y_sin = np.sin(x)
y_cos = np.cos(x)
plt.figure(figsize=(7, 6))
plt.subplot(2, 1, 1)#两行一列第一个
plt.plot(x, y_sin, color='royalblue')
plt.title('Sin')
plt.subplot(2, 1, 2)
plt.plot(x, y_cos, color='darkorange')
plt.title('Cos')
plt.tight_layout()
plt.show()
```

## 3. 直方图和 `np.histogram`

直方图展示数据分布。`np.histogram` 返回每个区间的计数和边界；`plt.hist` 可以直接把它绘制出来。

```python
rng = np.random.default_rng(42)
samples = rng.normal(loc=70, scale=10, size=500)
counts, edges = np.histogram(samples, bins=[40, 50, 60, 70, 80, 90, 100])
print('各区间计数：', counts)
print('区间边界：', edges)
plt.figure(figsize=(7, 4))
plt.hist(samples, bins=edges, edgecolor='black')
plt.title('成绩分布')
plt.xlabel('分数')
plt.ylabel('人数')
plt.show()
```

## 本节练习

使用 `x = np.linspace(0, 2*np.pi, 100)`：1. 绘制 `sin(x)` 和 `cos(x)` 到同一张图；2. 设置标题、坐标轴标签和图例；3. 再绘制 `sin(x) + 0.5*cos(2*x)`。

```python
x = np.linspace(0,2*np.pi,100)
plt.figure(figsize=(7, 4))
y = np.sin(x)
z = np.cos(x)
plt.plot(x,y,label='sin',c='red')
plt.plot(x,z,label='cos',c='blue')
plt.plot(x,np.sin(x)+0.5*np.cos(2*x),c='green',label='sin(x)+0.5cos(2x)')
plt.legend()
plt.title('对比')
plt.xlabel('x')
plt.ylabel('red-sin(x) blue-cos(x)')
plt.grid(True)
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)
plt.figure(figsize=(7, 4))
plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.plot(x, np.sin(x) + 0.5 * np.cos(2 * x), label='组合曲线')
plt.title('三条函数曲线')
plt.xlabel('x')
plt.ylabel('函数值')
plt.grid(True)
plt.legend()

plt.show()
```

### 小结

NumPy 生成数值数组，Matplotlib 将数组转换为图形。绘图时注意 x、y 长度一致、标题和单位清楚，并使用图例区分多条曲线。
