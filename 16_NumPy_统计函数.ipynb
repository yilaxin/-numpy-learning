# 第 16 课：NumPy 统计函数

NumPy 的统计函数可以沿整个数组或指定轴计算最小值、最大值、平均值、中位数、百分位数、方差和标准差。理解 `axis` 是本课重点：二维数组中，`axis=0` 按列统计，`axis=1` 按行统计。

```python
import numpy as np

print('NumPy 版本：', np.__version__)
```

## 1. 最小值、最大值和极差

`amin`/`min` 返回最小值，`amax`/`max` 返回最大值，`ptp` 返回最大值减最小值。

```python
scores = np.array([[80, 92, 75], [88, 79, 95]])

print('最小值：', np.amin(scores))
print('最大值：', np.amax(scores))
print('极差：', np.ptp(scores))
 
print('按列最小值：', np.min(scores, axis=0))
print('按行最大值：', np.max(scores, axis=1))
```

## 2. 平均值、中位数和百分位数

`mean` 是算术平均值，`median` 是中间位置的值，`percentile(a, q)` 表示第 `q` 百分位。`axis` 可让统计分别作用于行或列。

```python
data = np.array([10, 20, 30, 40, 50])

print('平均值：', np.mean(data))
print('中位数：', np.median(data))
print('第 25 百分位：', np.percentile(data, 25))
print('第 75 百分位：', np.percentile(data, 75))

print('scores 各科平均分：', np.mean(scores, axis=0))
print('scores 各位同学平均分：', np.mean(scores, axis=1))
```

## 3. 加权平均

`np.average` 可以通过 `weights` 为不同数据设置权重。`returned=True` 时会同时返回加权平均值和权重总和。

```python
values = np.array([80, 90, 70])
weights = np.array([0.2, 0.5, 0.3])
weighted, weight_sum = np.average(values, weights=weights, returned=True)

print('加权平均：', weighted)
print('权重总和：', weight_sum)
```

## 4. 方差和标准差

方差 `var` 衡量数据偏离平均值的程度；标准差 `std` 是方差的平方根。默认 `ddof=0` 计算总体统计量；样本统计通常使用 `ddof=1`。

```python
measurements = np.array([2, 4, 4, 4, 5, 5, 7, 9])

print('平均值：', np.mean(measurements))
print('总体方差：', np.var(measurements))
print('总体标准差：', np.std(measurements))
print('样本方差：', np.var(measurements, ddof=1))#这是样本是/（n-1）
print('样本标准差：', np.std(measurements, ddof=1))
```

## 5. 含 NaN 数据的统计

普通统计函数遇到 `np.nan` 通常会得到 `nan`。需要忽略缺失值时，使用 `nanmean`、`nanmedian`、`nanstd` 等函数。`keepdims=True` 会保留被统计的轴，便于后续广播。

```python
incomes = np.array([1000, np.nan, 1200, 1100])

print('普通平均值：', np.mean(incomes))
print('忽略 NaN 的平均值：', np.nanmean(incomes))

column_mean = np.mean(scores, axis=0, keepdims=True)
print('保留维度的列平均值：', column_mean)
print('形状：', column_mean.shape)
```

## 本节练习

给定 `temps = np.array([[18, 20, 22], [25, 27, 26], [30, 28, 29]])`，输出：

1. 全部数据的最小值、最大值和极差；2. 每一天的平均温度；3. 每个测量时刻的平均温度；4. 全部数据的中位数和标准差。

```python
temps = np.array([[18, 20, 22], [25, 27, 26], [30, 28, 29]])
print(np.min(temps))
print('max',np.max(temps))
print('ptp',np.ptp(temps))
print(np.mean(temps,axis=1))
print(np.mean(temps,axis=0))
print(np.median(temps))
print(np.std(temps))
```

```python
print('最小值：', np.min(temps))
print('最大值：', np.max(temps))
print('极差：', np.ptp(temps))
print('每天平均温度：', np.mean(temps, axis=1))
print('各测量时刻平均温度：', np.mean(temps, axis=0))
print('中位数：', np.median(temps))
print('标准差：', np.std(temps))
```
