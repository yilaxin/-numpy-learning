# 第十一课：NumPy 数组操作

本课依据菜鸟教程的 [NumPy 数组操作](https://www.runoob.com/numpy/numpy-array-manipulation.html) 整理，涵盖变形、换轴、增减维、连接、分割以及元素增删去重。

```python
import numpy as np
print("NumPy 版本：", np.__version__)
```

## 1. 修改形状：`reshape()`

新形状的元素总数必须与原数组相同。`-1` 可以让 NumPy 自动推算一个维度。`reshape()` 通常返回共享内存的视图，但不要依赖它永远不复制。

```python
a = np.arange(12)
b = a.reshape(3, 4)
c = a.reshape(2, -1)
print(b)
print("自动推算：", c.shape)
print("a 与 b 共享内存：", np.shares_memory(a, b))
```

## 2. 展平：`ravel()` 与 `flatten()`

两者都变成一维：`ravel()` 尽量返回视图，`flatten()` 总是返回副本。需要明确独立数据时用 `flatten()`。

```python
matrix = np.arange(8).reshape(2, 4)
raveled = matrix.ravel()
flattened = matrix.flatten()
print("C 顺序：", matrix.ravel(order="C"))
print("F 顺序：", matrix.ravel(order="F"))
print("ravel 共享内存：", np.shares_memory(matrix, raveled))
print("flatten 共享内存：", np.shares_memory(matrix, flattened))
```

## 3. 转置与交换轴

二维数组的 `.T` 或 `transpose()` 交换行列。多维数组中，`transpose(axes)` 可以重新排列全部轴；`swapaxes()` 只交换两个轴；`moveaxis()` 把指定轴移动到新位置，通常比旧式 `rollaxis()` 更直观。

```python
m = np.arange(12).reshape(3, 4)
print("原数组 shape：", m.shape)
print("转置 shape：", m.T.shape)
print(m.T)

cube = np.arange(24).reshape(2, 3, 4)
print(cube)
print("cube：", cube.shape)

print("transpose(2,0,1)：", cube.transpose(2, 0, 1).shape)
print(cube.transpose(2, 0, 1))

print("swapaxes(0,2)：", np.swapaxes(cube, 0, 2).shape)
print(cube)

print("moveaxis(0,-1)：", np.moveaxis(cube, 0, -1).shape)
print(cube)
```

## 4. 增加与删除长度为 1 的轴

`expand_dims()` 插入新轴；`squeeze()` 删除长度为 1 的轴。指定 `axis` 能避免误删其他轴。

```python
v = np.array([1, 2, 3])
row = np.expand_dims(v, axis=0)
column = np.expand_dims(v, axis=1)
print("原形状：", v.shape)
print("行向量：", row.shape)
print("列向量：", column.shape)
x = np.zeros((1, 3, 1))
print("squeeze 全部：", np.squeeze(x).shape)
print("只删 axis=0：", np.squeeze(x, axis=0).shape)
```

## 5. 沿已有轴连接：`concatenate()`

沿哪个轴连接，该轴长度可以不同；其他轴的长度必须一致。连接不会增加维数。

```python
left = np.array([[1, 2], [3, 4]])
right = np.array([[5, 6], [7, 8]])
along_rows = np.concatenate([left, right], axis=0)
along_cols = np.concatenate([left, right], axis=1)
print("axis=0：\n", along_rows, along_rows.shape)
print("axis=1：\n", along_cols, along_cols.shape)
```

## 6. 沿新轴堆叠：`stack()`

`stack()` 会增加一个新轴，因此输入数组形状必须完全一致。`vstack()` 和 `hstack()` 是常用便捷函数，但处理一维数组时规则不同，学习时优先关注结果 `shape`。

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("stack axis=0：\n", np.stack([a, b], axis=0), np.stack([a, b], axis=0).shape)
print("stack axis=1：\n", np.stack([a, b], axis=1), np.stack([a, b], axis=1).shape)
print("vstack：\n", np.vstack([a, b]))
print("hstack：", np.hstack([a, b]))
```

## 7. 分割数组

`split()` 要求平均分割时能够整除；`array_split()` 允许不均匀分割。`vsplit()` 沿第 0 轴，`hsplit()` 沿第 1 轴。

```python
a = np.arange(10)
print("按位置 3、7 分割：", np.split(a, [3, 7]))
print("不能整除时 array_split：", np.array_split(a, 3))#均匀分成三份
m = np.arange(16).reshape(4, 4)
print(m)
print("上下分割：", [part.shape for part in np.vsplit(m, 2)])
print("左右分割：", [part.shape for part in np.hsplit(m, 2)])
```

## 8. `resize()` 与 `reshape()` 不同

`reshape()` 要求元素总数不变；`np.resize()` 可改变元素总数，变大时会循环重复原数据。它返回新数组，不等同于数组方法 `arr.resize()` 的原地行为。

```python
small = np.array([1, 2, 3, 4])
print(np.resize(small, (3, 3)))
```

## 9. 添加、插入和删除

`append()`、`insert()`、`delete()` 都返回新数组，不会修改原数组。未提供 `axis` 时，输入会先被展平。反复追加大量数据效率较低，通常应先收集数据再一次性创建数组。

```python
m = np.array([[1, 2], [3, 4]])
print("追加一行：\n", np.append(m, [[5, 6]], axis=0))
print("追加一列：\n", np.append(m, [[9], [9]], axis=1))
print("第 2 列前插入：\n", np.insert(m, 1, 0, axis=1))
print("删除第 1 行：\n", np.delete(m, 0, axis=0))
print("原数组不变：\n", m)
```

## 10. 去重与计数：`unique()`

`np.unique()` 默认返回排序后的唯一值，还可以返回首次出现位置、反向映射和计数。

```python
labels = np.array(["A", "B", "A", "C", "B", "A"])
unique_values, inverse, counts = np.unique(labels, return_inverse=True, return_counts=True)
print("唯一值：", unique_values)
print("计数：", counts)#出现了几次
print("反向映射：", inverse)
print("可还原：", np.array_equal(unique_values[inverse], labels))
print(unique_values[inverse])
```

## 本课小结

- `reshape()` 改形状；`ravel()` 尽量返回视图；`flatten()` 返回副本。
- `.T`、`transpose()`、`swapaxes()`、`moveaxis()` 调整轴。
- `expand_dims()` 增轴，`squeeze()` 删长度为 1 的轴。
- `concatenate()` 沿已有轴连接；`stack()` 沿新轴堆叠。
- `split()` 严格均分，`array_split()` 允许不均分。
- 增删函数通常返回新数组；`unique()` 可去重并计数。
