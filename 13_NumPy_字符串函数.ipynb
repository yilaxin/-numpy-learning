```python
import numpy as np

print("NumPy 版本：", np.__version__)
```

## 1. 创建字符串数组

NumPy 会为普通字符串数组选择统一的 Unicode dtype。例如 `<U6` 表示每个元素最多保存 6 个 Unicode 字符。

```python
languages = np.array(["Python", "NumPy", "SQL"])

print(languages)
print("数组类型：", type(languages))
print("元素类型：", languages.dtype)
print("形状：", languages.shape)
```

### 固定长度字符串的截断风险

数组创建后，dtype 的字符长度通常已经固定。把更长的字符串直接赋进去可能被静默截断。需要保存更长字符串时，创建新的数组或显式使用足够长的 dtype。

```python
short_words = np.array(["cat", "dog"])
print("原数组：", short_words, short_words.dtype)

short_words[0] = "elephant"
print("赋值后：", short_words)

long_words = np.array(["cat", "dog"], dtype="U20")
long_words[0] = "elephant"
print("预留足够长度：", long_words, long_words.dtype)
```

## 2. 大小写转换

`np.char.lower()`、`upper()`、`capitalize()` 和 `title()` 会对数组中的每个字符串执行对应操作，并返回新数组。

```python
names = np.array(["alice smith", "BOB JONES", "cHaRlIe brown"])

print("原数据：", names)
print("全部小写：", np.char.lower(names))
print("全部大写：", np.char.upper(names))
print("仅首字符大写：", np.char.capitalize(names))
print("每个单词首字母大写：", np.char.title(names))
```

## 3. 字符串拼接与重复

`np.char.add()` 对应位置逐元素拼接，并支持广播；`np.char.multiply()` 对字符串逐元素重复。

```python
first_names = np.array(["Alice", "Bob", "Carol"])
last_names = np.array(["Smith", "Jones", "Lee"])

full_names = np.char.add(np.char.add(first_names, " "), last_names)
labels = np.char.add("用户：", full_names)

print("姓名：", full_names)
print("添加统一前缀：", labels)
print("重复字符串：", np.char.multiply(["-", "*="], [8, 3]))
```

## 4. 对齐与填充

`center()`、`ljust()` 和 `rjust()` 可以把每个字符串填充到指定宽度。它们适合生成简单的文本表格或编号。

```python
items = np.array(["A", "NumPy", "Python"])

print("居中：", np.char.center(items, 10, fillchar="-"))
print("居中：", np.char.center(items, 10, fillchar=" "))
print("左对齐：", np.char.ljust(items, 10, fillchar="."))
print("右对齐：", np.char.rjust(items, 10, fillchar="0"))
print("数字补零：", np.char.zfill(["7", "42", "315"], 5))
```

## 5. 清除两端字符

- `strip()`：处理两端
- `lstrip()`：只处理左端
- `rstrip()`：只处理右端

不指定字符时删除空白。指定字符时，它表示一个字符集合，而不是完整前缀或后缀。

```python
raw = np.array(["  apple  ", "  banana", "orange  "])
codes = np.array(["##A01##", "#B02", "C03###"])

print("清理空白：", np.char.strip(raw))
print("只清理左侧：", np.char.lstrip(raw))
print("只清理右侧：", np.char.rstrip(raw))
print("清理两端的 #：", np.char.strip(codes, "#"))
```

## 6. 替换、查找和计数

- `replace()`：替换子字符串
- `find()`：返回首次出现位置，找不到时返回 `-1`
- `count()`：统计子字符串出现次数

```python
texts = np.array(["data science", "database", "metadata", "python"])

print("替换：", np.char.replace(texts, "data", "DATA"))
print("data 的位置：", np.char.find(texts, "data"))#每个单词中的开始位置，没有就写-1
print("字母 a 的数量：", np.char.count(texts, "a"))
print("包含 data：", texts[np.char.find(texts, "data") >= 0])
```

## 7. 字符串判断与布尔筛选

`startswith()`、`endswith()`、`isdigit()`、`isalpha()` 等函数返回布尔数组，可以直接用于高级索引。

```python
files = np.array(["report.csv", "image.png", "data.csv", "README.md"])
csv_mask = np.char.endswith(files, ".csv")

print("判断结果：", csv_mask)
print("CSV 文件：", files[csv_mask])

values = np.array(["123", "12.5", "abc", "007"])
print("纯数字字符串：", values[np.char.isdigit(values)])
print("纯字母字符串：", values[np.char.isalpha(values)])
```

## 8. 分割字符串

`np.char.split()` 会分别拆分每个字符串。每个结果是一个 Python 列表，因此整体通常是 `object` 数组，而不是自动组成规则二维矩阵。`splitlines()` 按换行符拆分。
np.char.split() 默认按空白字符分割，包括：
- 普通空格
- 多个连续空格
- 制表符 \t
- 换行符 \n

```python
sentences = np.array(["NumPy is fast", "arrays are useful"])
parts = np.char.split(sentences)


print("拆分结果：", parts)
print("第一个句子的单词：", parts[0])

domains = np.array(["www.example.com", "docs.python.org"])
print("按点号拆分：", np.char.split(domains, sep="."))
print("按行拆分：", np.char.splitlines(["第一行\n第二行", "A\r\nB"]))
```

## 9. 小心 `np.char.join()` 的含义

`np.char.join(separator, string)` 会在**每个字符串的字符之间**插入分隔符。它不是把数组中的多个字符串合并为一个字符串。

要把整个数组连接起来，可以先转为列表，再使用 Python 的 `separator.join(...)`。

```python
print("字符间插入：", np.char.join("-", ["abc", "XYZ"]))

tags = np.array(["numpy", "python", "data"])
print("连接整个数组：", " | ".join(tags.tolist()))#tolist   numpy   数组抓换为列表
```

## 10. 编码与解码

`encode()` 将 Unicode 字符串编码为字节串，`decode()` 将字节串恢复为 Unicode。编码和解码必须使用匹配的编码名称。

```python
text = np.array(["NumPy", "数据"])
encoded = np.char.encode(text, "utf-8")
decoded = np.char.decode(encoded, "utf-8")

print("原字符串：", text, text.dtype)
print("UTF-8 字节：", encoded, encoded.dtype)
print("解码结果：", decoded, decoded.dtype)
print("恢复成功：", np.array_equal(text, decoded))
```

## 11. 实用流程：组合多个清洗步骤

字符串函数可以嵌套：先清除空白，再统一大小写，最后筛选或替换。每一步通常返回新数组，原数组不会被修改。

```python
raw_tags = np.array(["  Python", "NUMPY  ", " data-science ", "NumPy"])
clean_tags = np.char.lower(np.char.strip(raw_tags))
numpy_mask = np.char.find(clean_tags, "numpy") >= 0

print("原数据：", raw_tags)
print("清洗后：", clean_tags)
print("包含 numpy：", clean_tags[numpy_mask])
print("原数组未改变：", raw_tags)
```

## 本课小结

1. NumPy 字符串数组使用统一的固定长度 dtype，直接赋值过长字符串可能被截断。
2. `numpy.char` 中的函数对数组逐元素操作。
3. 大小写转换：`lower`、`upper`、`capitalize`、`title`。
4. 清洗和替换：`strip`、`replace`。
5. 查找与筛选：`find`、`count`、`startswith`、`endswith`。
6. `split()` 的每个结果是列表；`char.join()` 是在字符之间插入分隔符。
7. `encode()` 和 `decode()` 用于 Unicode 与字节串转换。
