```python
import tempfile
from pathlib import Path
import numpy as np

work = Path(tempfile.mkdtemp())
```

## 题目 1：数组属性和筛选

创建 `scores = [[78, 92, 85], [88, 76, 95], [90, 84, 89]]`，输出 shape、每位同学平均分、每科最高分，以及总分不低于 260 的同学行。

```python
scores = np.array([[78,92,85],[88,76,95],[90,84,89]])
print(scores.shape)
print(np.mean(scores,axis=1))
print(np.max(scores,axis=0))
print(scores[np.sum(scores,axis=1)>=260])
```

```python
scores = np.array([[78, 92, 85], [88, 76, 95], [90, 84, 89]])
print('shape：', scores.shape)
print('每位同学平均分：', scores.mean(axis=1))
print('每科最高分：', scores.max(axis=0))
print('总分不低于 250：\n', scores[scores.sum(axis=1) >= 250])
```

## 题目 2：排序和统计

将学生编号 `['A', 'B', 'C']` 按平均分从高到低排列，并输出平均分的中位数和标准差。

```python
students = np.array(['A','B','C'])
print(students[np.argsort(np.mean(scores,axis=1))][::-1])#np.argsort() 默认按从小到大排列。想按平均分从大到小，可以在索引后面加 [::-1]，把顺序反过来：
print(np.median(np.mean(scores,axis=1)))
print(np.std(np.mean(scores,axis=1)))
```

```python
students = np.array(['A', 'B', 'C'])
averages = scores.mean(axis=1)
order = np.argsort(averages)[::-1]
print('排名：', students[order])
print('平均分：', averages[order])
print('中位数：', np.median(averages))
print('标准差：', np.std(averages))
```

## 题目 3：解线性方程组

使用 `np.linalg.solve` 解方程组：`2x+y=5`、`x+3y=6`，并验证结果。

```python
A = np.array([[2., 1.], [1., 3.]])
b = np.array([5., 6.])
x = np.linalg.solve(A,b)
print(np.linalg.solve(A,b))
print(A@x)
```

```python
A = np.array([[2., 1.], [1., 3.]])
b = np.array([5., 6.])
solution = np.linalg.solve(A, b)
print('解：', solution)
print('验证：', A @ solution)
```

## 题目 4：保存并恢复

将 `scores` 保存到 `.npy` 文件，再读回并确认读回数组与原数组相等。

```python
file = work / 'scores.npy'
np.save(file, scores)
restored = np.load(file)
print('恢复结果：\n', restored)
print('是否相等：', np.array_equal(scores, restored))
```
