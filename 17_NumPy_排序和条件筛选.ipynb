# 第 17 课：NumPy 排序和条件筛选

本课学习 `sort`、`argsort`、`argmin`、`argmax`、`where`、`nonzero` 和 `extract`。它们分别用于排列数据、找到位置以及按条件选择数据。

```python
import numpy as np

print('NumPy 版本：', np.__version__)
```

## 1. `sort` 排序

`np.sort` 返回排序后的副本，不改变原数组。二维数组中可以用 `axis=0` 按列排序，用 `axis=1` 按行排序；`axis=None` 会先展平再排序。数组方法 `a.sort()` 则会原地修改数组。

```python
a = np.array([5, 2, 8, 1, 3])

print('原数组：', a)
print('升序副本：', np.sort(a))
print('原数组仍然是：', a)

matrix = np.array([[3, 1, 2], [0, 5, 4]])
print(matrix)
print('按行排序：\n', np.sort(matrix, axis=1))
print('按列排序：\n', np.sort(matrix, axis=0))

a.sort()
print('原地排序后：', a)
```

## 2. `argsort` 返回排序索引

`argsort` 不直接返回数据，而是返回从小到大排列所需的下标。用这些下标索引原数组，就能同时重排相关数据。`kind="stable"` 可保留相等元素的原相对顺序。

```python
names = np.array(['小王', '小李', '小张', '小赵'])
scores = np.array([88, 95, 95, 88])
order = np.argsort(scores, kind='stable') 

print('排序索引：', order)
print('按成绩升序：', names[order])
print('对应成绩：', scores[order])
print('最高分同学：', names[np.argmax(scores)])
```

## 3. 找到极值位置

`argmin` 和 `argmax` 返回最小值、最大值的索引。对二维数组指定 `axis` 时，返回每一行或每一列的索引。`partition` 可以快速找出第 k 小附近的元素，但它不保证整体有序。

```python
m = np.array([[8, 3, 9], [4, 7, 1]])

print('全局最小值位置：', np.argmin(m))
print('全局最大值位置：', np.argmax(m))
print('每列最小值所在行：', np.argmin(m, axis=0))
print('每行最大值所在列：', np.argmax(m, axis=1))
print('第二小元素在分区后的数组：', np.partition(m.ravel(), 1))#索引为1 的是3，就是把3放在sort后正确的位置，但是其它元素是随机的，只能保证分块有序，即左边的比3小，右边的比3大
```

## 4. 布尔索引和 `where`

比较运算会生成布尔数组，可以直接筛选元素。`np.where(condition)` 返回满足条件的索引；`np.where(condition, x, y)` 则按条件在 `x` 和 `y` 之间选择。

```python
values = np.array([12, 5, 18, 3, 20])

mask = values >= 10
print('布尔条件：', mask)
print('大于等于 10：', values[mask])
print('满足条件的索引：', np.where(mask))
print('分类标签：', np.where(values >= 10, '达标', '未达标'))
```

## 5. `nonzero`、`extract` 与组合条件

`nonzero` 返回非零元素的索引，`extract` 返回满足布尔条件的元素。组合条件时要给每个条件加括号，并使用 `&`（且）或 `|`（或），不能使用 Python 的 `and`、`or`。

```python
data = np.array([0, 4, -2, 0, 7, 9])

print('非零索引：', np.nonzero(data))
print('正数：', np.extract(data > 0, data))
print('大于 0 且小于 8：', data[(data > 0) & (data < 8)])
print('小于 0 或大于 8：', data[(data < 0) | (data > 8)])
```

## 本节练习

给定商品名和价格数组：

1. 按价格从低到高输出商品名；2. 找出最贵商品及其价格；3. 筛选价格不低于 200 的商品；4. 用 `where` 将价格分为“高价”和“普通”。

```python
products = np.array(['笔记本', '耳机', '鼠标', '键盘', '显示器'])
prices = np.array([4999, 299, 129, 399, 1599])

keys = np.argsort(prices,kind='stable')
print(products[keys])
print(np.max(prices))
print(products[np.argmax(prices)])
keys_ =  prices>=200
print( products[keys_])
print(np.where(prices>1000,'高价','普通'))
```

```python

order = np.argsort(prices)
print('价格升序商品：', products[order])
expensive = np.argmax(prices)
print('最贵商品：', products[expensive], prices[expensive])
print('价格不低于 20 的商品：', products[prices >= 20])
print('价格标签：', np.where(prices >= 1000, '高价', '普通'))
```

### 小结

`sort` 排数据，`argsort` 排索引；`argmin`/`argmax` 找位置；布尔索引、`where` 和 `nonzero` 负责条件筛选。实际项目中，先用索引排序，再同步重排名称、价格等关联数组，能避免数据错位。
