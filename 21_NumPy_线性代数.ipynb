# 第 21 课：NumPy 线性代数

`np.linalg` 集中了常用线性代数运算，包括点积、矩阵乘法、行列式、逆矩阵和解线性方程组。

```python
import numpy as np
```

## 1. 点积、内积和矩阵乘法

`dot` 对一维数组计算点积；`vdot` 会先展平并对复数取共轭；`inner` 计算最后一个轴上的内积；二维矩阵乘法推荐使用 `@` 或 `matmul`。

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])
print('dot：', np.dot(u, v))
print('vdot：', np.vdot(u, v))
print('inner：', np.inner(u, v))
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print('矩阵乘法：\n', A @ B)
```

## 2. 行列式、逆矩阵和范数

`np.linalg.det` 计算方阵行列式，`inv` 计算可逆矩阵的逆，`norm` 计算向量或矩阵范数。矩阵接近奇异时不应直接求逆，解方程通常使用 `solve`。

```python
M = np.array([[1., 2.], [3., 4.]])
M_inv = np.linalg.inv(M)
print('行列式：', np.linalg.det(M))
print('逆矩阵：\n', M_inv)
print('M @ inv(M)：\n', M @ M_inv)
print('向量范数：', np.linalg.norm(np.array([3, 4])))#范数为元素的模长
```

## 3. 解线性方程组

方程组 `AX = b` 可以用 `np.linalg.solve(A, b)` 求解，比手动计算 `inv(A) @ b` 更直接且通常更稳定。

```python
A = np.array([[1., 1., 1.], [0., 2., 5.], [2., 5., -1.]])
b = np.array([6., -4., 27.])
x = np.linalg.solve(A, b)
print('解：', x)
print('验证 A @ x：', A @ x)
```

## 4. 特征值和最小二乘

`eig` 返回方阵的特征值和特征向量；`lstsq` 可以在数据不完全满足方程时求最小二乘解。

```python
Q = np.array([[2., 0.], [0., 3.]])
eigenvalues, eigenvectors = np.linalg.eig(Q)
print('特征值：', eigenvalues)
print('特征向量：\n', eigenvectors)
X = np.array([[1., 1.], [1., 2.], [1., 3.]])
y = np.array([2., 2.9, 4.2])
coef, *_ = np.linalg.lstsq(X, y, rcond=None)
print('最小二乘系数：', coef)
```

## 5. 秩



```python
Z = np.random.randint(0,10,(10,10))
rank = np.linalg.matrix_rank(Z)
print(rank)
```

## 本节练习

给定 `A = [[2, 1], [1, 3]]`、`b = [5, 6]`：计算行列式、逆矩阵，并用 `solve` 求解 `Ax=b`，最后验证 `A @ x`。

```python
A = np.array([[2,1],[1,3]])
b = np.array([5,6])
print(np.linalg.det(A))
print(np.linalg.inv(A))
x = np.linalg.solve(A,b)
print(x)
print(A @ x == b)
```

```python
A = np.array([[2., 1.], [1., 3.]])
b = np.array([5., 6.])
x = np.linalg.solve(A, b)
print('det(A)：', np.linalg.det(A))
print('inv(A)：\n', np.linalg.inv(A))
print('x：', x)
print('验证：', A @ x)
```

### 小结

`dot`/`inner` 处理点积，`@` 处理矩阵乘法，`det`/`inv`/`solve` 用于矩阵分析和方程求解。下一课将学习数组文件的保存与读取。
