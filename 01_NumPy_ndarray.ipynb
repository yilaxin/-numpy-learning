# NumPy 第一课：`ndarray` 对象

`ndarray` 是 NumPy 最核心的数据结构，可以理解为“支持多维结构和高效计算的数组”。

关键特点：
- 元素通常具有相同的数据类型
- 可以是一维、二维或更多维
- 下标从 `0` 开始
- 支持整个数组一起计算
- 常用属性包括 `shape`、`ndim`、`size` 和 `dtype`

## 1. 导入 NumPy

```python
import numpy as np

print("NumPy 版本：", np.__version__)
```

## 2. 创建一维数组

`as np` 是约定俗成的简写。`np.array()` 可以把 Python 列表转换为 NumPy 数组。

```python
a = np.array([1, 2, 3])

print(a)
print(type(a))

python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])
print(type(python_list))
print(type(numpy_array))
```

## 3. 创建二维数组

二维数组可以理解为“行 × 列”。访问元素时通常使用 `b[行下标, 列下标]`。

```python
b = np.array([
    [1, 2],
    [3, 4]
])

print(b)
print("第 0 行第 0 列：", b[0, 0])
print("第 0 行第 1 列：", b[0, 1])
print("第 1 行第 0 列：", b[1, 0])
```

## 4. 数组的重要属性

| 属性 | 含义 | 示例 |
|---|---|---|
| `ndim` | 有几个维度 | `2` |
| `shape` | 每个维度有多长 | `(2, 2)` |
| `size` | 一共有几个元素 | `4` |
| `dtype` | 元素的数据类型 | 如 `int64` |

```python
print("数组：")
print(b)
print("维数 ndim：", b.ndim)
print("形状 shape：", b.shape)
print("元素数量 size：", b.size)
print("元素类型 dtype：", b.dtype)
```

## 5. 指定数据类型

`dtype` 可以指定数组元素的类型。数组中的元素会尽量统一为同一类型。

```python
c = np.array([1, 2, 3], dtype=complex)
print(c)
print(c.dtype)

d = np.array([1, 2, 3], dtype=np.float64)
print(d)
print(d.dtype)
```

## 6. 不同类型混合时的转换

```python
e = np.array([1, 2.5, 3])
print(e)
print(e.dtype)

f = np.array([1, 2, "hello"])
print(f)
print(f.dtype)
```

## 7. `ndmin` 最小维度

`ndmin=2` 表示要求结果至少是二维数组。

```python
g = np.array([1, 2, 3, 4, 5], ndmin=2)

print(g)
print("维数：", g.ndim)
print("形状：", g.shape)
```

## 8. NumPy 数组为什么适合计算？

Python 列表的 `* 2` 是重复列表；NumPy 数组的 `* 2` 是让每个元素都乘以 2。

```python
python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])

print(python_list * 2)
print(numpy_array * 2)
```
