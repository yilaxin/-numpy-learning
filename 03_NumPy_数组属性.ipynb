## 1. 一次查看主要属性

数组的维数也称轴数。二维数组有两个轴：`axis=0` 对应行方向，`axis=1` 对应列方向。

```python
a = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int32)
print(a)
print("ndim：", a.ndim)
print("shape：", a.shape)
print("size：", a.size)
print("dtype：", a.dtype)
print("itemsize：", a.itemsize, "字节")
print("nbytes：", a.nbytes, "字节")
```

| 属性 | 含义 | 本例结果 |
|---|---|---|
| `ndim` | 轴的数量 | `2` |
| `shape` | 每个轴的长度 | `(2, 3)` |
| `size` | 元素总数 | `6` |
| `dtype` | 元素类型 | `int32` |
| `itemsize` | 单个元素字节数 | `4` |
| `nbytes` | 全部元素字节数 | `24` |

关系：`size == np.prod(shape)`，`nbytes == size * itemsize`。

```python
scalar = np.array(7)
vector = np.array([1, 2, 3])
matrix = np.array([[1, 2], [3, 4]])
cube = np.arange(24).reshape(2, 3, 4)
# np.arange(24) 先产生一维的 24 个数
print(cube)
for name, x in [("标量", scalar), ("向量", vector), ("矩阵", matrix), ("三维数组", cube)]:
    print(f"{name:6} ndim={x.ndim}, shape={x.shape}, size={x.size}")
```

## 2. `shape` 与 `reshape()`

`shape` 是元组。`reshape()` 改变观察数据的形状，但元素总数必须保持一致。推荐使用 `reshape()`，比直接给 `shape` 赋值更清楚。

```python
original = np.arange(12)
reshaped = original.reshape(3, 4)
print("原数组：", original, original.shape)
print("新形状：\n", reshaped)
print("shape：", reshaped.shape)
print("是否可能共享内存：", np.shares_memory(original, reshaped))
```

`reshape()` 中可以有一个维度写成 `-1`，让 NumPy 自动推算。

```python
x = np.arange(24)
print(x.reshape(2, -1).shape)
print(x.reshape(2, 3, -1).shape)
```

## 3. 理解 `axis`

`axis` 表示运算时压缩哪个轴。二维数组中：

- `axis=0`：压缩行，得到每一列的结果
- `axis=1`：压缩列，得到每一行的结果

```python
scores = np.array([[80, 90, 70], [85, 95, 75]])
print(scores)
print("每列之和（axis=0）：", scores.sum(axis=0))
print("每行之和（axis=1）：", scores.sum(axis=1))
print("全部元素之和：", scores.sum())
```

## 4. 复数的 `real` 与 `imag`

复数数组可以分别取得实部和虚部。

```python
z = np.array([1 + 2j, 3 + 4j, 5 - 6j])
print("复数：", z)
print("实部：", z.real)
print("虚部：", z.imag)
```

## 5. `flags` 与内存布局

常用标志：`C_CONTIGUOUS` 表示按 C 风格连续存储，`F_CONTIGUOUS` 表示按 Fortran 风格连续存储，`OWNDATA` 表示数组拥有自己的数据，`WRITEABLE` 表示可写。初学阶段会查看即可。

C_CONTIGUOUS：按 C 风格连续存储，即最后一维变化最快。可以理解为按“行”依次放进内存。

F_CONTIGUOUS：按 Fortran 风格连续存储，即第一维变化最快。可理解为按“列”依次放进内存。

OWNDATA：这个数组是否拥有自己的一块数据内存。

WRITEABLE：数组内容能否修改。

```python
base = np.arange(6).reshape(2, 3)
transposed = base.T#转置
print(base)
print(transposed)
for name, x in [("原数组", base), ("转置后", transposed)]:
    print(name, x.shape)
    print("  C 连续：", x.flags.c_contiguous)
    print("  F 连续：", x.flags.f_contiguous)
    print("  拥有数据：", x.flags.owndata)#这里为false，因为数据都与np.arange(6)共享共享。没有独立的内存空间
    print("  可写：", x.flags.writeable)
```

可以把数组设为只读来防止误修改。下面用副本演示，避免影响前面的变量。

```python
readonly = a.copy()
readonly.flags.writeable = False
print("可写：", readonly.flags.writeable)
#readonly[0, 0] = 99  # 取消注释会触发 ValueError
```

## 本课小结

- `ndim` 看维数，`shape` 看每个轴长度，`size` 看元素总数。
- `itemsize` 是单个元素内存，`nbytes` 是全部元素内存。
- `reshape()` 改形状时元素总数不变，并且常与原数组共享内存。
- `axis` 指被压缩的轴。
- `flags` 提供内存布局和可写性信息。

## 本节练习

创建一个包含 1 到 24 的 `int16` 数组，并整理成 `(2, 3, 4)`：

1. 输出 `ndim`、`shape`、`size`、`itemsize`、`nbytes`。
2. 验证 `nbytes == size * itemsize`。
3. 分别计算 `axis=0`、`axis=1`、`axis=2` 的和，并观察结果形状。

```python
 
practice = np.arange(1, 25, dtype=np.int16).reshape(2, 3, 4)
# 请在这里完成练习
print(practice.ndim)
print(practice.shape)
print(practice.size)
print(practice.nbytes)
print(practice.itemsize)

print(practice.nbytes == practice.size * practice.itemsize)

print(list(range(3)))

for i in list(range(3)):
    print(practice.sum(axis=i)) 
    print()
print(practice)
```

## 参考答案

```python
print("ndim：", practice.ndim)
print("shape：", practice.shape)
print("size：", practice.size)
print("itemsize：", practice.itemsize)
print("nbytes：", practice.nbytes)
print("关系成立：", practice.nbytes == practice.size * practice.itemsize)
for axis in range(practice.ndim):
    result = practice.sum(axis=axis)
    print(f"axis={axis}，结果 shape={result.shape}：\n{result}")
```
