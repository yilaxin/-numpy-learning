```python
# 第 7 题：每行减去该行平均值
a = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

a_mean = np.mean(a,axis=1).reshape(3,1)
print(a_mean)
print(a-a_mean)
```

```python
# 第 8 题：按第 2 列升序排列整行
a = np.array([
    [101, 88, 20],
    [102, 65, 19],
    [103, 95, 21],
    [104, 72, 18]
])
print(a[:,1])
order = np.argsort(a[:,1]) 

a_ = a[order]
print(a_)


```
