# 第七课：NumPy 切片和索引

本课依据菜鸟教程的 [NumPy 切片和索引](https://www.runoob.com/numpy/numpy-indexing-and-slicing.html) 整理，并补充负索引、多维切片、维度保留和视图共享等实用知识。

学习目标：

- 使用整数索引取得单个元素
- 掌握 `start:stop:step` 切片语法
- 对二维和三维数组进行索引
- 理解切片通常是视图而不是副本

```python
import numpy as np
print("NumPy 版本：", np.__version__)
```

## 1. 一维数组索引

索引从 `0` 开始。负索引从末尾开始计数：`-1` 是最后一个元素。

```python
a = np.arange(10)
print("数组：", a)
print("a[0]：", a[0])
print("a[5]：", a[5])
print("a[-1]：", a[-1])
print("a[-3]：", a[-3])
```

## 2. 切片语法 `start:stop:step`

- `start`：起始索引，包含
- `stop`：停止索引，不包含
- `step`：步长

三者都可以省略。`a[2:7:2]` 会取得索引 2、4、6。

```python
a = np.arange(10)
print("数组：", a)
print("a[2:7:2]：", a[2:7:2])
print("a[:4]：", a[:4])
print("a[6:]：", a[6:])
print("a[::2]：", a[::2])
print("a[::-1]：", a[::-1])
```

`slice(2, 7, 2)` 与 `2:7:2` 表达相同的切片。在需要动态构造切片时，`slice()` 很有用。

```python
s = slice(2, 7, 2)
print(a[s])
print(a[2:7:2])
```

## 3. 二维数组索引

二维数组一般写作 `array[行, 列]`。逗号左边选择第 0 轴，右边选择第 1 轴。

```python
m = np.arange(1, 13).reshape(3, 4)
print(m)
print("第 2 行第 3 列：", m[1, 2])
print("第 1 行：", m[0, :])
print("第 2 列：", m[:, 1])
print("右下角：", m[-1, -1])
```

每个轴都可以使用切片。`m[0:2, 1:3]` 表示取前两行与第 2～3 列的交叉区域。

```python
print("原数组：\n", m)
print("前两行、第 2 到第 3 列：\n", m[0:2, 1:3])#：不包含尾部元素
print("每隔一行、每隔一列：\n", m[::2, ::2])
```

## 4. 整数索引会减少维度

`m[1]` 使用整数选中一行，结果是一维；`m[1:2]` 使用切片选中一个行区间，仍保留二维。机器学习中需要保持 `(1, n)` 形状时尤其要注意。

```python
# m = m.T这样可以对列进行操作
row_1d = m[1]
row_2d = m[1:2]
print("m[1]：", row_1d, "shape=", row_1d.shape)
print("m[1:2]：\n", row_2d, "shape=", row_2d.shape)
```

## 5. 省略号 `...`

`...` 代表中间所有未明确写出的轴。维度较高时可减少大量冒号。对于二维数组，`m[..., 1]` 等同于 `m[:, 1]`。

```python
cube = np.arange(24).reshape(2, 3, 4)
print("cube：\n", cube)
print("cube shape：", cube.shape)
print("所有页、所有行、第 2 列：\n", cube[..., 1])
print("第 1 页的所有内容：\n", cube[0, ...])
print("最后一列：\n", cube[..., -1])
```

## 6. 补充：用 `None` / `np.newaxis` 增加维度

它们不会增加数据，只是改变数组的形状。常用于把一维数组转换成行向量或列向量。

```python
v = np.array([10, 20, 30])
row = v[:,np.newaxis]#根据新建维度相对：的位置来给出新维度的形状
column = v[None,np.newaxis,]#所以用None和np.newaxis是一样的，都是增加一个维度
print("原数组 shape：",v, v.shape)
print("行向量：\n", row, "shape=", row.shape)
print("列向量：\n", column, "shape=", column.shape)
```

## 7. 重要：基本切片通常返回视图

NumPy 切片通常与原数组共享内存。修改切片可能同时修改原数组。这与 Python 列表切片会创建新列表不同。需要独立数据时使用 `.copy()`。

```python
original = np.arange(6)
print("原数组：", original)
view = original[1:4]
print("切片：", view)
print("共享内存：", np.shares_memory(original, view))
view[0] = 99
print("修改切片后，原数组：", original)
independent = original[1:4].copy()
independent[0] = -1
print("修改副本后，原数组：", original)
```

## 8. 使用切片批量赋值

给切片赋一个标量时，NumPy 会把该值应用到选中的所有位置。后面会学习这背后的广播机制。

```python
board = np.zeros((4, 5), dtype=np.int32)
board[1:3, 1:4] = 7
board[:, 0] = -1
print(board)
```

## 本课小结

- 索引从 0 开始，负索引从末尾开始。
- 切片格式为 `start:stop:step`，不包含 `stop`。
- 多维数组用逗号分隔各轴的选择。
- 整数索引通常减少一个维度，切片通常保留维度。
- `...` 表示中间未写出的所有轴。
- 基本切片通常与原数组共享内存，需要独立数据时使用 `.copy()`。
